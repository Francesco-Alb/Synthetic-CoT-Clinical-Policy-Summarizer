### Synthetic Data Generation Pipeline: Model Strategy

### 1. The Generators (Diversity of Thought)
To avoid "model collapse" in synthetic datasets, we use distinct architectures to provide a wider distribution of linguistic styles and logic paths:

* **Model A: `gemini/gemma-3-27b-it`**
    * **Role:** The Anchor. Gemma 3 provides state-of-the-art instruction following and reasoning in a 27B parameter dense format, serving as the high-efficiency baseline for the dataset.
* **Model B: `meta-llama/llama-4-scout-17b-16e-instruct`**
    * **Role:** The Creative. As a 16-expert Mixture-of-Experts (MoE) model, it generates a broader range of edge cases and nuanced variations than dense models.

### 2. The Judge (Quality Control)
The "Judge" must significantly exceed the intelligence of the generators to accurately identify hallucinations or low-quality outputs.

* **The Judge: `openai/gpt-oss-120b`**
    * **Role:** The Auditor. With 120B parameters, it possesses the "world knowledge" and reasoning depth required to audit the outputs of the 17B–32B generators.

---

### Summary Table

| Role | Model Choice | Key Strength for Synthetic Data |
| :--- | :--- | :--- |
| **Gen 1** | `Gemma 3 27B` | SOTA instruction following and reasoning. |
| **Gen 2** | `llama-4-scout` | Diversity of output via 16-expert MoE. |
| **Judge** | `gpt-oss-120b` | Superior auditing for filtering/scoring. |

### Downsample

````python
# Select n samples per category within a quantile range (token efficiency)
p25, p90 = trials_df['full_len'].quantile([.25,.90])
downsampled_trials_df = trials_df.query("full_len >= @p25 and full_len <= @p90")
downsampled_trials_df['search_term_used'].value_counts()
print(f"--Retaining only texts with condition: {p90} >= length >= {p25} -- ")

# Sample n samples per each search_term_used
n_sample = 30
downsampled_trials_df = (
    downsampled_trials_df
    .groupby('search_term_used')
    .apply(lambda x: x.sample(n_sample, random_state=config.env.seed))
    .reset_index(drop=True)
    )

# Check
downsampled_trials_df['search_term_used'].value_counts()
````


>I used a quantile-based filtering strategy ($p25$ to $p90$) to optimize token efficiency while ensuring a balanced representation across four medical domains. I implemented a fallback logic to augment short summaries with detailed trial descriptions to ensure the model had sufficient evidence for reasoning.

### Define Pipeline

````python
# Drop row from dictionary if the silver_checkpoint exists and it has already got a row with the same nctId
if os.path.isfile(config.data.checkpoint_csv_path) and config.data.preprocess_data:
    silver_checkpoint = pd.read_csv(config.data.checkpoint_csv_path)
    existing_nctIds = silver_checkpoint['nctId'].to_list()
    processed_data = [item for item in processed_data if item['nctId'] not in existing_nctIds]
    print(f"Discarding {len(existing_nctIds)} existing_nctIds.")
    print(f"Remaining rows: {len(processed_data)}")
else:
    print("No silver_checkpoint found.")
````

````python
with Pipeline(name="CoT-Pipeline") as cot_pipeline:
    loader = LoadDataFromDicts(name="load_data", batch_size=1, data=processed_data if config.data.preprocess_data else raw_data)

    reasoning_steps = []
    for llm in config.model.llm_configs:
        model_id = llm.model_name.split('/')[-1].replace('-', '_').replace('.', '_')

        step = ThrottledTextGeneration(
            name=f"generate_{model_id}",
            llm=llm,
            columns=["prompt"] if config.data.preprocess_data else ["instruction", "document", "target_audience"],
            # columns=["nctId", "prompt"] if config.data.preprocess_data else ["instruction", "document", "target_audience"],
            template="{{ prompt }}" if config.data.preprocess_data else "INSTRUCTION: {{ instruction }}\n DOCUMENT: {{ document }}\nAUDIENCE: {{ target_audience }}",
            output_mappings={
                "generation":"generation",
                "model_name":"model_name"
                },
            input_batch_size=1,
        )
        reasoning_steps.append(step)

    group_responses = GroupColumns(
        columns=["generation", "model_name"],
        output_columns=["generations", "model_names"],
    )

    saver = CheckpointStep(name="checkpoint_save", file_path=config.data.checkpoint_csv_path)

    if config.model.add_judge_step and not config.model.decouple_judge:
        ranker = UltraFeedback(
            name="ultrafeedback_ranking",
            llm=config.model.judge_model,
            aspect="overall-rating",
            input_mappings={
                "instruction": "prompt",
                "generations": "generations",
            },
            output_mappings={
                "model_name": "judge_model"
            }
        )

        loader >> reasoning_steps >> group_responses >> ranker >> saver

    else:
        loader >> reasoning_steps >> group_responses >> saver
````

### **Architectural Rationale: Optimized Teacher Generation**

To ensure high-fidelity "Silver" data generation without hitting API rate limits, the pipeline is configured for **high-TPM stability** rather than raw speed.

#### **1. Model Selection (The Pareto Mix)**
* **Gemini-3-27B (Google):** 15,000 TPM limit.
* **Llama-4-Scout-17B (Groq):** 30,000 TPM limit.

#### **2. Throttling & Batch Math**
* **Input Batch Size:** `1` (Processing one trial at a time to prevent token "bursting").
* **Delay:** `10.0 seconds` (6 requests per minute).
* **Expected Token Usage:** ~1,000 tokens per request.
* **Calculated TPM:** $6 \text{ RPM} \times 1,000 \text{ Tokens} = 6,000 \text{ TPM Total}$.
* **Safety Margin:**
    * **Google:** ~60% headroom (6k used vs 15k limit).
    * **Groq:** ~80% headroom (6k used vs 30k limit).
